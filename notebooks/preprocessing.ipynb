```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/usgs_inventory_missing_cleaned.csv")

print("Cleaned dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Cell 19: Define initial modelling features

target = "Response"
group_column = "FireName"

full_features = [
    "StormAccumulation",
    "StormDuration",
    "PeakI15",
    "PeakI30",
    "PeakI60",
    "1YearStormI15",
    "1YearStormI30",
    "1YearStormI60",
    "RainfallRatio15",
    "RainfallRatio30",
    "RainfallRatio60",
    "HM23",
    "dNBR/1000",
    "KF",
    "UpstreamArea",
    "DaysSinceFire",
    "YearAfterFire"
]

reduced_features = [
    "StormAccumulation",
    "StormDuration",
    "PeakI60",
    "RainfallRatio60",
    "HM23",
    "dNBR/1000",
    "KF",
    "UpstreamArea",
    "DaysSinceFire",
    "YearAfterFire"
]

print("Full feature count:", len(full_features))
print("Reduced feature count:", len(reduced_features))
```

```python
# Cell 20: Prepare feature matrix and target

X = df[full_features].copy()
y = df[target].copy()

groups = df[group_column].copy()

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Number of wildfire groups:", groups.nunique())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Cell 21: Check data quality before modelling

print("Missing values in X:")
print(X.isnull().sum())

print("\nInfinite values in X:")
print(np.isinf(X.select_dtypes(include=np.number)).sum())

print("\nMissing values in target:", y.isnull().sum())
print("Missing wildfire groups:", groups.isnull().sum())
```

```python
# Cell 22: Missing-value percentages in our initial features

feature_missing = (
    X.isnull().mean() * 100
).sort_values(ascending=False)

feature_missing = feature_missing[feature_missing > 0]

print(feature_missing.round(2))
```
