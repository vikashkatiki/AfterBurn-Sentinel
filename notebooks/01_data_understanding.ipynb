```python

# Cell 1: Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully!")
```

```python

# Cell 2: Load the USGS dataset

data_path = "../data/raw/USGS_RG_PFDF_Inventory_v1.csv"

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

# Cell 3: Understand dataset structure

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python

# Cell 4: Statistical summary

df.describe(include="all").T
```

```python

# Cell 6: Missing values

# In this dataset, -9999 represents missing/unavailable values
df.replace(-9999, np.nan, inplace=True)

missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100
})

missing = missing.sort_values(
    by="Missing Percentage",
    ascending=False
)

missing
```

```python

# Cell 7: Correctly replace all missing-value markers

df.replace([-9999, "-9999"], np.nan, inplace=True)

print("Missing-value markers cleaned.")

print("\nDataset information after cleaning:")
df.info()
```

```python

# Cell 8: Updated missing-value analysis

missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100
})

missing = missing[missing["Missing Count"] > 0]
missing = missing.sort_values(
    by="Missing Percentage",
    ascending=False
)

missing
```

```python

# Cell 9: Updated statistical summary

df.describe(include="all").T
```

```python

# Cell 19: Save cleaned working dataset

import os

os.makedirs("../data/processed", exist_ok=True)

df.to_csv(
    "../data/processed/usgs_inventory_missing_cleaned.csv",
    index=False
)

print("Cleaned working dataset saved successfully!")
```
